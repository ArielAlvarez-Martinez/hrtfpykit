# Dummy SimpleFreeFieldHRIR Example

This notebook shows how to create a dummy **SimpleFreeFieldHRIR** SOFA dataset and populate its core variables. Each step explains *what* we do and *why* it is required.

## 1) Imports
We need NumPy for numeric arrays and `SOFA` to create and edit the dataset.

```python
import numpy as np
from hrtfpykit import SOFA
```

## 2) Create a dummy SOFA dataset
We create a **SimpleFreeFieldHRIR** dummy with explicit dimension sizes.

- `M` = number of measurements (directions)
- `R` = receivers (ears)
- `N` = time samples
- `C` = coordinate components

These sizes must match the shapes of the data we will write next.

```python
dummy = SOFA.create_dummy(
    "SimpleFreeFieldHRIR",
    dim_sizes={"M": 100, "N": 256, "R": 2, "C": 3},
)
```

## 3) Build dummy data arrays
The SimpleFreeFieldHRIR convention stores the impulse response in **Data.IR**.
So we create:
- `Data.IR`: shape `(M, R, N)`
- `Data.SamplingRate`: scalar sampling rate in Hz

These are the minimum required pieces to represent an HRIR in the time domain.

```python
dummy_ir = np.random.randn(100, 2, 256)
dummy_sr = np.array([48000.0])
```

## 4) Write the arrays into the SOFA variables
We overwrite the default variables created by `create_dummy()` with our arrays.

This ensures the dataset contains valid numeric content and a sampling rate that
matches the data.

```python
dummy.modify_variable("Data.IR", dummy_ir)
dummy.modify_variable("Data.SamplingRate", dummy_sr)
```

## 5) Quick sanity checks
We check that shapes match the expected dimensions. This is a quick validation that our arrays
align with the dataset layout.

```python
vars_wrap = dummy.Variables
print("Data.IR shape (M, R, N):", vars_wrap.get("Data.IR").value.shape)
print("Data.SamplingRate shape:", vars_wrap.get("Data.SamplingRate").value.shape)
print(dummy.summary())
```

## 6) Save the dummy SOFA file
We save the in-memory dataset to disk so it can be reused or inspected later.
Use a dummy path for testing.

```python
dummy.save("dummy_output_hrir.sofa", overwrite=True)
```
