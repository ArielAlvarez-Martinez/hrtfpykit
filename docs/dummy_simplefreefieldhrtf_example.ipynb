# Dummy SimpleFreeFieldHRTF Example

This notebook shows how to create a dummy **SimpleFreeFieldHRTF** SOFA dataset and populate its core variables. Each step explains *what* we do and *why* it is required.

## 1) Imports
We need NumPy for numeric arrays and `SOFA` to create and edit the dataset.

```python
import numpy as np
from hrtfpykit import SOFA
```

## 2) Create a dummy SOFA dataset
We create a **SimpleFreeFieldHRTF** dummy with explicit dimension sizes.

- `M` = number of measurements (directions)
- `R` = receivers (ears)
- `N` = frequency bins
- `C` = coordinate components

These sizes must match the shapes of the data we will write next.

```python
dummy = SOFA.create_dummy(
    "SimpleFreeFieldHRTF",
    dim_sizes={"M": 100, "N": 256, "R": 2, "C": 3},
)
```

## 3) Build dummy data arrays
The SimpleFreeFieldHRTF convention stores the transfer function as **real** and **imaginary** parts.
So we create:
- `Data.Real`: shape `(M, R, N)`
- `Data.Imag`: shape `(M, R, N)`
- `N`: a frequency axis of length `N` in Hz

These are the minimum required pieces to represent an HRTF in the frequency domain.

```python
dummy_real = np.random.randn(100, 2, 256)
dummy_imag = np.random.randn(100, 2, 256)
dummy_n = np.linspace(0.0, 22050.0, 256)
```

## 4) Write the arrays into the SOFA variables
We overwrite the default variables created by `create_dummy()` with our arrays.

This ensures the dataset contains valid numeric content and a frequency axis that matches the data.

```python
dummy.modify_variable("Data.Real", dummy_real)
dummy.modify_variable("Data.Imag", dummy_imag)
dummy.modify_variable("N", dummy_n)
```

## 5) Quick sanity checks
We check that shapes match the expected dimensions. This is a quick validation that our arrays
align with the dataset layout.

```python
vars_wrap = dummy.Variables
print("Data.Real shape (M, R, N):", vars_wrap.get("Data.Real").value.shape)
print("Data.Imag shape (M, R, N):", vars_wrap.get("Data.Imag").value.shape)
print("N shape (frequency bins):", vars_wrap.get("N").value.shape)
print(dummy.summary())
```

## 6) Save the dummy SOFA file
We save the in-memory dataset to disk so it can be reused or inspected later.
Use a dummy path for testing.

```python
dummy.save("dummy_output.sofa")
```
